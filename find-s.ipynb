```python
import numpy as np 
instance = np.array(df)[:,:-1]
instance
```

```python
target = np.array(df)[:,-1]
target
```

```python
def train(ins,tar):
    #specific_h = ins[0].copy()
    for i,val in enumerate(tar):
        if val == 'YES':
            specific_h = ins[i].copy()
            break  
    for i,val in enumerate(ins):
        if tar[i] == 'YES':
            for x in range(len(specific_h)):
                if val[x] != specific_h[x]:
                    specific_h[x] = '?'
                else:
                    pass
    return specific_h   
   
```

```python
print("The final Hypothesis is:",train(instance,target))
```
